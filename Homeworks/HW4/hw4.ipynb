We could attempt linear regression using `sns.lmplot` as shown below, which hint at some relationships between temperature and proportional casual, but the plot is still fairly unconvincing. 

```python
sns.lmplot(data=bike, x="temp", y="prop_casual", hue="weekday", scatter_kws={"s": 20}, height=10)
plt.title("Proportion of Casual Riders by Weekday");
```

A better approach is to use local smoothing. The basic idea is that for each x value, we compute some sort of representative y value that captures the data close to that x value. One technique for local smoothing is "Locally Weighted Scatterplot Smoothing" or LOWESS. An example is below. The red curve shown is a smoothed version of the scatterplot.

```python
from statsmodels.nonparametric.smoothers_lowess import lowess
# Make noisy data
xobs = np.sort(np.random.rand(100)*4.0 - 2)
yobs = np.exp(xobs) + np.random.randn(100) / 2.0
sns.scatterplot(x = xobs, y = yobs, label="Raw Data")

# Predict 'smoothed' valued for observations
ysmooth = lowess(yobs, xobs, return_sorted=False)
sns.lineplot(x = xobs, y = ysmooth, label="Smoothed Estimator", color='red')
plt.legend();
```

```python
from statsmodels.nonparametric.smoothers_lowess import lowess

plt.figure(figsize=(9, 6))

# Sunday
sunday = bike[bike["weekday"] == "Sun"]
ysmooth_sun = lowess(sunday["prop_casual"], sunday["temp"], return_sorted=False)
sns.lineplot(x=sunday["temp"], y=ysmooth_sun, label="Sunday")

# Monday
monday = bike[bike["weekday"] == "Mon"]
ysmooth_mon = lowess(monday["prop_casual"], monday["temp"], return_sorted=False)
sns.lineplot(x=monday["temp"], y=ysmooth_mon, label="Monday")

# Tuesday
tuesday = bike[bike["weekday"] == "Tue"]
ysmooth_tue = lowess(tuesday["prop_casual"], tuesday["temp"], return_sorted=False)
sns.lineplot(x=tuesday["temp"], y=ysmooth_tue, label="Tuesday")

# Wednesday
wednesday = bike[bike["weekday"] == "Wed"]
ysmooth_wed = lowess(wednesday["prop_casual"], wednesday["temp"], return_sorted=False)
sns.lineplot(x=wednesday["temp"], y=ysmooth_wed, label="Wednesday")

# Thursday
thursday = bike[bike["weekday"] == "Thu"]
ysmooth_thu = lowess(thursday["prop_casual"], thursday["temp"], return_sorted=False)
sns.lineplot(x=thursday["temp"], y=ysmooth_thu, label="Thursday")

# Friday
friday = bike[bike["weekday"] == "Fri"]
ysmooth_fri = lowess(friday["prop_casual"], friday["temp"], return_sorted=False)
sns.lineplot(x=friday["temp"], y=ysmooth_fri, label="Friday")

# Saturday
saturday = bike[bike["weekday"] == "Sat"]
ysmooth_sat = lowess(saturday["prop_casual"], saturday["temp"], return_sorted=False)
sns.lineplot(x=saturday["temp"], y=ysmooth_sat, label="Saturday")


plt.xlabel("Temperature (°C)")
plt.ylabel("Proportion of Casual Riders")
plt.title("LOWESS Curves: Proportion of Casual Riders vs Temperature by Weekday")
plt.legend(title="Weekday", ncol=2)
plt.tight_layout()
plt.show()
```
